```python
####
#  prompt -> LLM -> Answer
#  -----------------------// GenAI
#
# Agentic AI
# ============
# LLM
#  |
#  |-> Understand Goal
#  |-> Decide Action
#  |-> Use tool
#  |-> Observe result
#  |-> Iterate if needed
#  |
#  |___ Answer

# GenAI = LLM
#  Vs
# Agentic = LLM + Tool

#      +------------------------+
#      |  Goal Understanding
#      |      |
#      |    Planning
#      |      |
#      |    Decide Next Action
#      +-------------------------+
#                |
#      ---------- -----------------|
#      |                           |
#      Direct Answer             Use Tool
#                                  |
#                                 tool_name
#                                  |
#                                 Result
#                                  |
#                                 Observation
#                                  |
#                                 ...

from langchain_groq import ChatGroq   # llm  - pre-defined classes

from langchain_core.tools import tool  # tool - userdefined - function

from langchain.agents import create_agent # agent - predefined function

```

```python
# langchain tool is decorator
# decorator - function - adding more features to an existing code
# @<decoratorName>
# def UserdefinedFunction():
#    ....... code logic


# Agent = LLM + tool + prompt
#         ---   ====   ------
#         |      |         |->prompt
#         |      |->initialize functionname(UserdefinedFunction)
#         |
# initialize llm object

# Invoke Agent
```

```python
# Extend to Agentic Style code
#  Agent = LLM + Tool + Prompt

# LLM+prompt <== GenAI
# Tool - only <== this is not AI code - normal function

from langchain_groq import ChatGroq
from langchain_core.tools import tool
from langchain.agents import create_agent


# 1. Create llm
llm_obj = ChatGroq(model="openai/gpt-oss-20b",api_key=myapi_key)


# 2. Create a tool
@tool
def mycalc(expression: str) -> str:
  """simple mathematic operation"""
  return str(eval(expression))

# 3. Create Agent = LLM + tool
agent_obj = create_agent(
    model = llm_obj,
    tools = [mycalc]
)

agent_obj.invoke({"messages":[{"role":"user","content":"Calculate 125 * 87 and explain the result"}]})

# HumanMessage <== user Input
# AIMessage <== llm response
# SystemMessage  <== prompt template
# ------------------// pre-defined class





```
